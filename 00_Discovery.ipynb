Xây dựng nhãn churn phù hợp với thương mại điện tử phi hợp đồng từ hành vi không mua lại.

So sánh ảnh hưởng của các ngưỡng không mua 90, 120 và 180 ngày.
Xây dựng feature RFM, giao dịch, review, giao hàng, thanh toán và sản phẩm mà không dùng thông tin tương lai.

So sánh Logistic Regression, Random Forest, XGBoost và LightGBM trong cùng một protocol.

Đánh giá bằng temporal validation và các metric phù hợp với dữ liệu mất cân bằng.

Giải thích dự đoán bằng Feature Importance và SHAP, sau đó đề xuất nhóm khách cần giữ chân.
